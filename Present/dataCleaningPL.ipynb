# Reusable Data Cleaning Pipeline

This notebook demonstrates a simple reusable data-cleaning pipeline using Python and pandas.

```python
# Step 1: Import library
import pandas as pd
```

```python
# Step 2: Create sample dirty dataset
data = {
    "Name": ["Santosh", "Pradeep", None, "Suyog", "Sahan"],
    "Salary": [60000, None, 61000, 78000, 80000],
    "City": ["Pokhara", "Kathmandu", "Pokhara", None, "Kathmandu"]
}

df = pd.DataFrame(data)
print("Original Data:")
df
```

```python
# Step 3: Create reusable cleaning functions

def remove_missing_names(df):
    return df.dropna(subset=["Name"])

def fill_missing_salary(df):
    df["Salary"] = df["Salary"].fillna(df["Salary"].mean())
    return df

def fill_missing_city(df):
    df["City"] = df["City"].fillna("Unknown")
    return df
```

```python
# Step 4: Create pipeline function

def clean_data_pipeline(df):
    # work on a copy up front to prevent chained-assignment warnings
    df = df.copy()
    df = remove_missing_names(df)
    df = fill_missing_salary(df)
    df = fill_missing_city(df)
    return df
```

```python
# Step 5: Apply pipeline

clean_df = clean_data_pipeline(df)

print("Cleaned Data:")
clean_df
```
