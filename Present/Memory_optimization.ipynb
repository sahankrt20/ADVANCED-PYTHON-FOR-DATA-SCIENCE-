# MEMORY OPTIMIZATION USING PYTHON
Memory optimization in Python means reducing RAM usage and making programs efficient, which is especially important for large datasets, ML pipelines, and high-performance system

# UNDERSTAND HOW PYTHON USES MEMORY
Python stores everything as objects:
<ul>
<li>Each variable → object reference</li>

<li>Objects have overhead (type info, ref count, metadata)</li>
Example:

```python
a = 10
```

This is not just 10 — it's a full Python integer object (~28 bytes).

# USE SET INSTEAD OF LIST FOR MEMBERSHIP CHECKS

```python
# Faster + avoids duplicates
data1=[1,2,3,4,5,5,4,3]
data2=[1,2,3,4,5,5]
items=set(data1)
if data2[0] in items:
    print("Found")
```

# Generators Instead of Lists 

Lists store everything in memory.

Generators produce values on demand.

```python
nums = [x*x for x in range(1,100)]
nums
```

```python
nums = (x*x for x in range(100))
nums
```

# Use Iterators for File/Data Processing

Never load big files full

```python
data = open("file.txt").read().splitlines()
data
```

```python
with open("file.txt") as f:
    for line in f:
        print(line.strip())
```

# Use __slots__ for Classes

Default classes store attributes in dictionaries.

```python
class User:
    def __init__(self,name,age):
        self.name=name
        self.age=age
```

```python
class User:
    __slots__ = ("name","age")
    def __init__(self,name,age):
        self.name=name
        self.age=age
```

Reduces memory 40–60% per object.
